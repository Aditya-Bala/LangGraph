```python
from dotenv import load_dotenv
from langgraph.graph import StateGraph, START, END
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from typing import TypedDict
```

```python
load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id="openai/gpt-oss-20b",
    task="text-generation"
)

model = ChatHuggingFace(llm = llm)
```

```python
class llmmessage(TypedDict):
    title : str
    outline : str
    content : str
```

```python
def create_outline(state : llmmessage) -> llmmessage:

    title = state['title']

    prompt = f'Create a brief outline based on the given title, make it sound exciting, \n {title}'

    answer = model.invoke(prompt)

    state['outline'] = answer.content

    return state



def create_content(state : llmmessage) -> llmmessage:

    outline = state['outline']

    prompt = f'Create descriptive content for blog based on the given title, make it sound exciting, interesting and factual, \n {outline}'

    answer = model.invoke(prompt)

    state['content'] = answer.content
    return state
```

```python
graph = StateGraph(llmmessage)

graph.add_node("create_outline",create_outline)
graph.add_node("create_content",create_content)
```

```python
graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_content')
graph.add_edge('create_content',END)
```

```python
workflow = graph.compile()
```

```python
answer = workflow.invoke({'title' : "World war 2"})
```
