```python
from dotenv import load_dotenv
from langgraph.graph import StateGraph, START, END
from langchain_huggingface import HuggingFaceEmbeddings, HuggingFaceEndpoint, ChatHuggingFace

from typing import TypedDict
```

```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace

load_dotenv()

#llm = HuggingFaceEndpoint(
#    repo_id="mistralai/Mistral-7B-Instruct-v0.2",  # free model
 #   task='text-generation'
#)
llm = HuggingFaceEndpoint(
    repo_id="openai/gpt-oss-20b",
    task="text-generation"
)

model = ChatHuggingFace(llm = llm)
```

```python
class llmmessage(TypedDict):
    question: str
    answer: str
```

```python
def qa(state : llmmessage) -> llmmessage:
    question = state['question']

    prompt = f'Answer the following question {question}'

    state['answer'] = model.invoke(prompt).content

    return state
```

```python

graph = StateGraph(llmmessage)

graph.add_node('qa',qa)

graph.add_edge(START,'qa')
graph.add_edge('qa',END)

workflow = graph.compile()
```

```python
workflow.invoke({'question' : "What is stock market"})
```
